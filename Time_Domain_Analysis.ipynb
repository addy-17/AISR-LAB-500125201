```python
import numpy as np
import matplotlib.pyplot as plt

g = 9.81
L = 1.8
theta_0 = np.radians(50)

dt = 0.01
t = np.arange(0,10,dt) # Corrected: np.arrange to np.arange

omega_n = np.sqrt(g/L) # Corrected: l to L
theta_t = theta_0 * np.cos(omega_n * t)

N = len(t)
fft_vals = np.fft.fft(theta_t)
freqs = np.fft.fftfreq(N,dt)

pos_mask = freqs>=0 # Corrected: frqs to freqs
freqs = freqs[pos_mask]
fft_magnitude = np.abs(fft_vals[pos_mask]) * (2.0/N)

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.plot(t,np.degrees(theta_t))
plt.title("Time Domain")
plt.xlabel('time (s)')
plt.ylabel('angle (deg)')
plt.grid(True)



plt.subplot(1,2,2)
plt.plot(freqs,fft_magnitude)
plt.title("Frequency Domain")
plt.xlabel('frequency (Hz)')
plt.ylabel('magnitude')
plt.xlim(0,2)
plt.grid(True)

plt.tight_layout()
plt.show()
```
